Python Assignment – Functions

Write a Python program using separate functions for addition, subtraction, multiplication, and division.

```python


def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b


num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Addition =", add(num1, num2))
print("Subtraction =", subtract(num1, num2))
print("Multiplication =", multiply(num1, num2))
print("Division =", divide(num1, num2))
```

Write functions to convert Celsius to Fahrenheit, Fahrenheit to Celsius, and Celsius to Kelvin.

```python

def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32


def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9


def celsius_to_kelvin(celsius):
    return celsius + 273.15


c = float(input("Enter temperature in Celsius: "))
f = float(input("Enter temperature in Fahrenheit: "))

print("Temperature in Fahrenheit:",
      celsius_to_fahrenheit(c))

print("Temperature in Celsius:",
      fahrenheit_to_celsius(f))

print("Temperature in Kelvin:",
      celsius_to_kelvin(c))
```

Create a module named module_factorial.py and import it into another Python program

```python

def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact

import module_factorial

num = int(input("Enter a number: "))

if num < 0:
    print("Factorial is not possible for a negative number.")
else:
    result = module_factorial.factorial(num)
    print("Factorial of", num, "is", result)
```

Write a Python function to check whether a number is prime or not.

```python

def check_prime(num):
    if num <= 1:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


number = int(input("Enter a number: "))

if check_prime(number):
    print(number, "is a prime number.")
else:
    print(number, "is not a prime number.")
```

Write a Python function named string_reverse to reverse a string.

```python

def string_reverse(text):
    reversed_text = text[::-1]
    return reversed_text


word = input("Enter a string: ")

result = string_reverse(word)

print("Reversed string is:", result)
```
